```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np


```

```python
df_gapminder = px.data.gapminder()
display(df_gapminder.head())
```

```python
df_2007 = df_gapminder[df_gapminder['year'] == 2007]


fig = px.scatter(
    df_2007,
    x='gdpPercap',
    y='lifeExp',
    title="Expectativa de Vida vs PIB per capita(2007)"

)

fig.show()
```

```python
fig = px.scatter(
    df_2007,
    x='gdpPercap',
    y='lifeExp',
    color="continent",
    size="pop",
    hover_name="country",
    log_x=True,
    size_max=60,
    title="Expectativa de Vida vs PIB per capita(2007) - Melhorado"

)

fig.show()
```

```python
df_brasil = df_gapminder[df_gapminder['country'] == 'Brazil']

fig = px.line(
    df_brasil,
    x='year',
    y='lifeExp',
    title="Expectativa de vida no Brasil 1952-2007",
    markers=True
)

fig.show()
```

```python
paises_sulamericanos = [' Brazil', 'Argentina', 'Chile', 'Colombia', 'Peru']
df_sul = df_gapminder[df_gapminder['country'].isin(paises_sulamericanos)]


fig = px.line(
    df_sul,
    x='year',
    y='lifeExp',
    color='country',
    title="Evoluçao Da Expectativa de Vida na América Do Sul",
    markers=True
)

fig.show()
```

```python
df_americas_2007 = df_2007[df_2007['continent']=='Americas'].sort_values('pop', ascending=False)

fig = px.bar(
    df_americas_2007,
    x ="country",
    y ="pop",
    title=" Popualaçao dos países das américas (2007)",
    color="lifeExp",
    color_continuous_scale=px.colors.sequential.Viridis,

)

fig.show()

```

```python
df_tips = px.data.tips()
display (df_tips.head())
```

```python
fig = px.histogram(
    df_tips,
    x='total_bill',
    title="Distribuição do Valor Total da Conta",
    nbins= 30,
    marginal="box",
    color_discrete_sequence=['indianred']
)

fig.show()
```

```python
fig = px.box(
    df_tips,
    x="day",
    y="total_bill",
    color="smoker",
    title="Valor Total da Conta por Dia da Semana",
    points="all",
    category_orders={"day": ["Thur", "Fri", "Sat", "Sun"]}
)

fig.show()
```

```python
fig = px.violin(
    df_tips,
    x="day",
    y="total_bill",
    color="sex",
    box=True,
    points="all",
    title="Gráfico Violin da Conta por Dia da Semana",
    category_orders={"day": ["Thur", "Fri", "Sat", "Sun"]}
)

fig.show()
```

```python
pop_continente = df_2007.groupby('continent')['pop'].sum().reset_index()


fig = px.pie(
    pop_continente,
    values='pop',
    names='continent',
    title='distribuiçao da populaçao Mundial  por Continente',
    hole=0.0
)

fig.update_traces(textposition='inside', textinfo='percent+label')

fig.show()
```

```python
fig = px.treemap(
    df_2007,
    path=[px.Constant('Mundo'), 'continent', 'country'],
    values='pop',
    color='lifeExp',
    color_continuous_scale='RdBu',
    title='Treemap: Expectativa de Vida '
)
fig.update_layout(margin=dict(t=50, l=25, r=25, b=25))
fig.show()
```

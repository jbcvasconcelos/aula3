```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
sns.set_theme(style="whitegrid", palette="pastel")
```

```python
tips = sns.load_dataset("tips")
```

```python
display(tips.head())
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x="total_bill", y="tip", data=tips)
```

```python
plt.title("relação entre valor da conta e gorjeta")
plt.show
```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
)

plt.title("relação contra gorjeta, separado po sexo")
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day",
    style="time",
    size="size",
    sizes=(20, 200),
    alpha=0.7,
)

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.title("Scatter Plot Multidimensional")
plt.tight_layout()
plt.show
```

```python
flights = sns.load_dataset("flights")
display (flights.head())

plt.figure(figsize=(12, 8))
sns.lineplot(
    data=flights,
    x="year",
    y="passengers",
    )
plt.title("evolução do numero de passageiros")
plt.show
```

```python
plt.figure(figsize=(10, 5))
sns.countplot(
    data=tips,
    x="day",
    hue="day",
    palette="pastel",
    legend=False
)

plt.title("quantidade de mesas atendidas por dia ")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    palette="muted",
)
plt.title("media valor da conta por dia, separado por sexo")
plt.ylabel("valor medio por conta")
plt.show()
#
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="smoker",
    palette="Set1",
)

plt.title("distribuição dos valores da conta por dia")
plt.show()
#
```

```python
plt.figure(figsize=(12, 6))
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    split=True,
    inner="quartile",
    palette="Set2",
)

plt.title("distribuição dos valores da conta por dia")
plt.show()
#
```

```python
tgl, axes = plt.subplots(1, 2, figsize=(12, 6))

sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    jitter=True,
    alpha=0.7,
    ax=axes[0],
)
axes[0].set_title("strip Plot (com Jitter)")

sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill",
    ax=axes[1],
)
axes[1].set_title("swarm Plot")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    inner=None,
    color=".8",
    alpha=0.5,
)

sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    size=3,
    color=".3",
    alpha=0.5,
)

plt.title("distribuição dos valores da conta por dia")
plt.show()
#
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,
    kde=True,
    color="purple",

)

plt.title("distribuição dos valores da conta por dia")
plt.xlabel("valor da conta")
plt.ylabel("frequencia")
plt.show()
#
```

```python
plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="time", # Corrected: changed 'heu' to 'hue' and added a comma
    fill=True,
    alpha=0.5,
    linewidth=2,

)

plt.title("distribuição dos valores da conta por dia")
plt.show()
```

Heatmap ( Mapa de Calor) matris de correlação, iris(media de flores)

```python
from numpy._core import numeric
iris = sns.load_dataset("iris")
display(iris.head())

correlacao = iris.corr(numeric_only=True)

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlacao,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f",
    linewidths=1,
    square=True,
)
plt.title("matriz de correlação do conjunto de dados Iris")
plt.show()
#
```

Pairplot (O Canivete Suiço da analise exploratoria

```python
sns.pairplot(
    data=iris,
    hue="species",
    palette="viridis",
    diag_kind="kde",
    markers=["o", "s", "D"]
)
plt.title("parplot do conjunto de dados Iris", y=1.02, fontsize=16)
plt.show()
#
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(16, 12))
fig.suptitle('Dashboard de Analise de Gorjetas' , fontsize=20, fontweight='bold')

sns.scatterplot(data=tips, x='total_bill', y='tip', hue= 'sex', ax=axes[0, 0])
axes[0, 0].set_title('dispersão conta vs gorjeta')

sns.histplot(data=tips, x='total_bill', kde=True, color='green', ax=axes[0, 1])
axes[0, 1].set_title('distribuição conta')

sns.boxplot(data=tips, x='day', y='total_bill', hue='day', palette='Set2', ax=axes[1, 0], legend=False)
axes[1, 0].set_title('Boxplot por Dia')

sns.barplot(data=tips, x='day', y='total_bill', hue='smoker', ax=axes[1, 1])
axes[1, 1].set_title('media por Refeição e Fumante')

plt.tight_layout()
plt.show()
```
